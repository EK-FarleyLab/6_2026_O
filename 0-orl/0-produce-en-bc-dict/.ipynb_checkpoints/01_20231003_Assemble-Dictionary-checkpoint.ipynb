# Create scripts to analyze data

```python
# a python script to analyze a chunk of the fastq file

pyscript='''

import easyMPRA_20231003 as em

out='01-assemble-dict/0-collapsed-reads-{date}-{fn}-part-{part}.tsv'

fq="{input_fq}"
date="{input_date}"
numTotalReads="{input_totreads}"

bn=fq.split('/')[-1].split('.fastq')[0]
outfn=out.replace('{fn}',bn).replace('{date}',str(date))

print('reading',bn,'...')
print('outputting to',outfn,'...')

read2count=em.dict_step1_raw_to_collapsed_reads(rawReadFile=fq,
                                                fileType='fastq',
                                                outfn=outfn,
                                                readCol0idx=None,
                                                numberReadsPerFile=100e6,
                                                fastqTotalLines=numTotalReads)
                                                    
'''
```

# write submission scripts for server

```python
# we use a server to analyze each file 

qsub_template='''#!/bin/bash

### The queue (-q) to which the job is being submitted to:
### *condo is the queue for when we need more resources than we own,
### *glean is the free queue based on open resources on TSCC,
### *home  is our home queue that only farley lab members can use
### The account (-A) will always be farley-group
#PBS -q condo
#PBS -A farley-group

### Specify the name of the job, output file, and error/log file;
### Generally, you can keep the same naming scheme across all files
### $PBS_JOBID embeds the job id in the script (helpful for debugging, but can be removed)
#PBS -N JOB
#PBS -o quickjob.$PBS_JOBID.log
#PBS -e quickjob.$PBS_JOBID.err

### Request resources for the job, and the walltime for the job to run 
### (note: ppn - processors per node; maximum ppn is 16)
### Generally try to run jobs with a single node and increase ppn
### 1 ppn roughly = 5 gb
#PBS -l nodes=1:ppn=32
#PBS -l walltime=8:00:00

### Exports all user environment variables to the job
#PBS -V

### Specify the email (can be multiple, separated by commas) to recieve 
### emails/diagnostic information about the job; -m sets the mail settings:
### (n: no mail sent, a: mail sent when job aborted, b: mail sent when job begins,
### e: mail sent when job terminates)
#PBS -M solvason@eng.ucsd.edu
#PBS -m abe

#######################################################
# !!!!! DIAGNOSTIC STATEMENTS BELOW THIS HEADER !!!!! #
#######################################################

# Print out diagnostic information to have within the output file,
# This information can be removed or kept as desired
echo $PBS_JOBID
echo $PBS_JOBNAME
echo $PBS_NODELIST
echo $PBS_O_WORKDIR
echo $PBS_USER

# Go to the directory from which the script was called before running any commands
cd $PBS_O_WORKDIR

##############################################
# !!!!! JOB COMMANDS BELOW THIS HEADER !!!!! #
##############################################

source ~/.bash_profile
source ~/.condainit
conda activate notebook

date

python {input_pyscript}

date
'''
```

```python
# get name of all pytho scrips to run to submit to server
pyscripts=!ls 01a-collapsed-reads-pyscripts/*.py
pyscripts=[i.split('/')[-1] for i in pyscripts]
pyscripts
```

```python

for pyscript in pyscripts:
    print(pyscript)
    
    with open ('01a-collapsed-reads-pyscripts/GO.'+pyscript.replace('.py',''),'w') as f: f.write(qsub_template.replace('{input_pyscript}',pyscript))
```

# Load enhancer order sheet

```python
# make python script to parse enhancer from bc

pyscript='''

import easyMPRA_20231003 as em

enOrderSheet='../otxa-scrambled-wt-and-ablated.tsv'
    
fn='{input_col_reads}'

print(f'reading {fn}...')
read2count=em.read2count_tsv_to_dict(fn)

print('trimming reads...')
read2count={read[:-5]:count for read,count in read2count.items()}

print('parsing barcodes...')
readsWithEnhancerAndBarcode=\
em.dict_step2SR_parse_en_bc_from_read(enOrderSheet=enOrderSheet,
                                      read2count=read2count,
                                      bcLength=30,
                                      relPositionsToLookForLinker=[-5,-4,-3,-2,-1,+1],
                                      linkerSeq='ACTAGAAGAGGAGAATGGTTTGC',
                                      outfn='01b-parsed-reads/1-parsed-enhancers-barcodes-{basename}.tsv',
                                      returnRevCompEn=True,
                                      returnRevCompBc=False)

print('assembling...')
bc2en2count=\
em.dict_step3_assemble_dict(readsWithEnhancerAndBarcode=readsWithEnhancerAndBarcode,
                            enOrderSheet=enOrderSheet,
                            outDictBasename='01c-raw-dicts/2_raw-dict-{basename}')
'''
```

```python
read2countfiles=!ls 01a-collapsed-reads/*.tsv
read2countfiles=[i for i in read2countfiles]
read2countfiles[:3]
```

```python

# make python script and server submission files for parsing enhancer/barcodes from reads

for fn in read2countfiles:
    # print(fn)
    basename=fn.replace('01a-collapsed-reads/0-collapsed-reads-','').split('.tsv')[0]
    # print(basename)
    
    pyscriptname=f'01b-parsed-reads-pyscripts/parse-barcodes-{basename}.py'
    with open(pyscriptname,'w') as f: f.write(pyscript.replace('{basename}',basename).replace('{input_col_reads}',fn))
    
    goscriptname=f'01b-parsed-reads-pyscripts/GO.parse-barcodes-{basename}'
    with open(goscriptname,'w') as f: f.write(qsub_template.replace('{input_pyscript}',pyscriptname))
```

# Combine all dictionaries to find unique barcodes

```python
fnlist=!ls 01b-parsed-reads/*.tsv

fnlist
```

```python
!head '01b-parsed-reads/1-parsed-enhancers-barcodes-20200224-OLS_Dict_i12_S2_L001_R1_001-part-1.tsv'
```

```python
# combine all dictionaries into a single python dictionary 

combined_bc2en2count={}
    
for fn in fnlist:
    print('reading',fn,'...')
    readsWithEnhancerAndBarcode=em.readsWithEnhancerAndBarcode_tsv_to_dict(fn)
    for en,bc,readcount in readsWithEnhancerAndBarcode:
        
        readcount=float(readcount)
        
        if bc not in combined_bc2en2count:
            combined_bc2en2count[bc]={}
        
        if en not in combined_bc2en2count[bc]:
            combined_bc2en2count[bc][en]=0
        
        combined_bc2en2count[bc][en]+=readcount
        
```

```python

minreads=1
ubc2en=\
em.dict_step4_filter_dictionary_for_unique_and_min_reads(bc2en2count=combined_bc2en2count,
                                                         minReads=minreads,
                                                         outDictBasename=f'01c-dicts//3-filt-dict',
                                                         bcTrimLength=None)
```
