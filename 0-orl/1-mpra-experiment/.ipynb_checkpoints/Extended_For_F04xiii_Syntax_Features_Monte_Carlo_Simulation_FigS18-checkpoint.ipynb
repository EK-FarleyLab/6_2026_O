```python
# the length of syntax features to use
# featlen=2
featlen=3
```

```python
iter2bin2percentEnsDiffAndPvalue={}
for iteration in range(numits):

    random.seed(iteration)

    ################################################
    # Define bin edges
    ################################################
    
    if featlen==2:
        l1,l2,  m1,m2,  h1,h2 = 0,2,  4,6,  8,11
    
    elif featlen==3:
        l1,l2,  m1,m2,  h1,h2 = 0,2,  4,6,  8,12
    
    low=pd.Interval( l1, l2, closed='left' )
    lm = pd.Interval( l2, m1, closed='left' )
    med=pd.Interval( m1, m2, closed='left' )
    mh =pd.Interval( m2, h1, closed='left' )
    high=pd.Interval(h1, h2, closed='left' )
    
    bin_edges = [l1, l2, m1, m2, h1,h2]
    
    ################################################
    # load syntax features
    ################################################
    
    fn='F05-Syntax-Features/6-real-and-imputed-motifs.pd.df.pickle'
    sigDF=pd.read_pickle(fn).reset_index()
    
    sigDF=sigDF.loc[sigDF['phrase-length']==featlen,:]
    
    phrases=[]
    for phrasekey in sigDF['phrase-key']:
        phrasekey=phrasekey.split('=')[-1]
        phrasekey=phrasekey.split('_')
        thisphrase=[]
        for i in phrasekey:
            try:
                i=int(i)
            except ValueError:
                pass
            thisphrase.append(i)
        phrases.append(tuple(thisphrase))
    sigDF['phrase']=phrases
    sigDF['percent-func']=sigDF['num-active-ens']/sigDF['num-total-seqs']
    sigDF.sort_values('percent-func',ascending=False).head(10)
    
    # generate syntax feature set with revcomps
    sigMotifsFR=set()
    for fwd in sigDF.phrase:
        rev=syn.revcomp(fwd,'ord_ori_spc')
        sigMotifsFR.add(fwd)
        sigMotifsFR.add(rev)
    
    len(sigMotifsFR)
    
    ################################################
    # load seq to syn feat mapping 
    ################################################
        
    syn2en2afflist=load_pickle_dict('F05-Syntax-Features-testset-analysis/syn2en2afflist.pydict.pickle')
    en2syn2afflist=load_pickle_dict('F05-Syntax-Features-testset-analysis/en2syn2afflist.pydict.pickle')
    
    ################################################
    # load mpra data
    ################################################
    
    fn='01-Calculate-Enhancer-Activity-Levels/02-generated-tables//endf.rescaledPctl_45_85.wtcollapsed.sumRnaDivSumDna-repSum.pd.df.pickle'
    df=pd.read_pickle(fn)
        
    #############
    # apply labels
    
    # discard na values
    nullValuesMpraDf=sum(df['wt-pool-sumR/sumD-sum-filtNa-Dnotnull-minActFilt0.5-log2-rescaled'].isna())
    df=df.loc[df['wt-pool-sumR/sumD-sum-filtNa-Dnotnull-minActFilt0.5-log2-rescaled'].notnull(),:]
    
    # count numnber of unclassified 
    u=sum((df['wt-pool-sumR/sumD-sum-filtNa-Dnotnull-minActFilt0.5-log2-rescaled']>0) & (df['wt-pool-sumR/sumD-sum-filtNa-Dnotnull-minActFilt0.5-log2-rescaled']<1))
    c=len(df)-u #number classified
    
    # call top half active and bottom half inert
    a=int(c/2)
    i=int(c/2)

    df=df.sort_values('wt-pool-sumR/sumD-sum-filtNa-Dnotnull-minActFilt0.5-log2-rescaled',ascending=False)
    df['alabel']=[1]*a + [np.NaN]*u + [0]*i
    
    # discard null values so you only have 1 and 0 left
    df=df.loc[df['alabel'].notnull(),:]
    
    ################################################
    # count number of syntax features in each sequence
    ################################################
    
    # determine number motifs in real data
    numMotifsCol=[]
    motifsContainedCol=[]
    for seq in df['wt-enhancer-seq']:
        allPossibleOrgs=list(en2syn2afflist[seq].keys())
        allSigMotifsInSeq = [i for i in allPossibleOrgs if i in sigMotifsFR]
    
        motifsContainedCol.append(allSigMotifsInSeq)
        numMotifsCol.append(len(allSigMotifsInSeq))
    
    df['number-sig-syntax-features']=numMotifsCol
    df['list-sig-syntax-features']=motifsContainedCol
        
    #######################################
    # bin real data
    #######################################
    
    # Create bins
    df['num-sig-syntax-features-bins'] = pd.cut(df['number-sig-syntax-features'], bins=bin_edges,right=False)
    print(df['num-sig-syntax-features-bins'].value_counts())
        
    ##############################################################################
    # plot
    ##############################################################################
    
    fig,ax=quickfig(5,5)
    
    x=0
    
    # plot all
    numens=sum(df['alabel'])
    numinert=len(df['wt-pool-sumR/sumD-sum-filtNa-Dnotnull-minActFilt0.5-log2-rescaled'].dropna())-numens
    tot=numens+numinert
    numens=numens/tot
    numinert=numinert/tot
    
    ax.bar(x=x,height=numinert,bottom=0,color=cb['orangenature'])
    ax.bar(x=x,height=numens,bottom=numinert,color=cb['green'])
    
    x+=1
    
    REALDATA_bin2percentfunc={}
    
    for bi in [low,lm,med,mh,high]:
    
    
        theseRows=df['num-sig-syntax-features-bins']==bi
        tot=sum(theseRows)
    
        numens=sum(df.loc[theseRows,'alabel'])
        numinert=tot-numens
    
        if tot==0:
            continue
            
        REALDATA_bin2percentfunc[bi]=(numens,tot)
    
        numens=numens/tot
        numinert=numinert/tot
    
    
        ax.bar(x=x,height=numinert,bottom=0,color=cb['orangenature'])
        ax.bar(x=x,height=numens,bottom=numinert,color=cb['green'])
    
        x+=1
        
    ax.set_xticks([0,1,2,3,4,5],labels=['all',low,lm,med,mh,high])
    ax.set_ylabel("Fraction of Elements")
    ax.set_title('real data')
        
    plt.show()
    
    ############################################################
    # simulate syntax features in elements
    ############################################################
    
    Ef=a
    Ei=i # subsample inerts to be same probability as enhancer. this removes artifact    
    
    ##################################################################
    # generate elements 
    ##################################################################
    
    element=1
    element2func={}
    func2elements={1:set(),0:set()}
    element2features={}
    
    for i in range(Ef):        
        element2func[element]=1
        
        func2elements[1].add(element)
        element2features[element]=set()
        element+=1
        
    for i in range(Ei):        
        element2func[element]=0
        
        func2elements[0].add(element)
        element2features[element]=set()
        element+=1

    print('num func ',len(func2elements[1]))
    print('num inert ',len(func2elements[0]))
    
    ##################################################################
    # assign features to elements
    ##################################################################
    
    nElementsWithFeatThreshold_min=0
    nActiveElementsWithFeatThreshold_max=np.inf
    
    x=[]
    y=[]
    
    feat=1
    for Fa,_,Fi in sigDF['real-data']:
    
        nElementsWithFeat=Fa+Fi
        
        x.append(Fa)
        y.append(Fi)
        
        funcElementsChosen  = random.sample(list(func2elements[1]), Fa)
        
        inertElementsChosen = random.sample(list(func2elements[0]), Fi)
    
        for ei in funcElementsChosen:
            element2features[ei].add(feat)
    
        for ei in inertElementsChosen:
            element2features[ei].add(feat)
    
        feat+=1
    
    ##################################################################
    # make dataframe
    ##################################################################
    
    c2v={c:[] for c in ['element','label','nfeatures']}
    
    for element,featlist in element2features.items():
    
        c2v['element'].append(element)
        c2v['label'].append(element2func[element])
        c2v['nfeatures'].append(len(element2features[element]))
    
    simdf=pd.DataFrame(c2v)
    
    ##################################################################
    # bin simulated data by number of syntax feats
    ##################################################################
    
    simdf['nfeatures-bin']=pd.cut(simdf['nfeatures'], bin_edges,right=False)
    
    
    simdf['nfeatures-bin'].value_counts()
    print(len(simdf))
    print(simdf['nfeatures-bin'].value_counts())
    
    ###################################################
    # determine counts
    ###################################################

    lowSyntaxRows=  simdf['nfeatures-bin']==low
    lmSyntaxRows =  simdf['nfeatures-bin']==lm
    medSyntaxRows=  simdf['nfeatures-bin']==med
    mhSyntaxRows =  simdf['nfeatures-bin']==mh
    highSyntaxRows= simdf['nfeatures-bin']==high
    
    # get numbers
    
    el = sum(simdf.loc[lowSyntaxRows,'label'])  # enhancers low syntax
    nl = sum(lowSyntaxRows)
    il = nl-el
    
    elm = sum(simdf.loc[lmSyntaxRows,'label'])  # enhancers low syntax
    nlm = sum(lmSyntaxRows)
    ilm = nlm-elm
    
    em = sum(simdf.loc[medSyntaxRows,'label'])  # enhancers low syntax
    nm = sum(medSyntaxRows)
    im = nm-em
    
    emh = sum(simdf.loc[mhSyntaxRows,'label'])  # enhancers low syntax
    nmh = sum(mhSyntaxRows)
    imh = nmh-emh
    
    eh = sum(simdf.loc[highSyntaxRows,'label'])  # enhancers low syntax
    nh = sum(highSyntaxRows)
    ih = nh-eh
    
    
    PERMDATA_bin2percentfunc={low:(el,nl),lm:(elm,nlm),med:(em,nm),mh:(emh,nmh),high:(eh,nh)}
    
    # ###################################################
    # # plot simulated data
    # ###################################################
    
    fig,ax=plt.subplots(1,figsize=(5,5),dpi=200)
    
    ax.bar(x=0,height=100*Ei/(Ei+Ef),bottom=0,color=cb['orangenature'],width=.7)
    ax.bar(x=0,height=100*Ef/(Ei+Ef),bottom=100*Ei/(Ei+Ef),color=cb['green'],width=.7)
    
    if nl>0:
        ax.bar(x=1,height=100*il/nl,bottom=0,color=cb['orangenature'],width=.7)
        ax.bar(x=1,height=100*el/nl,bottom=100*il/nl,color=cb['green'],width=.7)
    
    if nlm>0:
        ax.bar(x=2,height=100*ilm/nlm,bottom=0,color=cb['orangenature'],width=.7)
        ax.bar(x=2,height=100*elm/nlm,bottom=100*ilm/nlm,color=cb['green'],width=.7)
    
    if nm>0:
        ax.bar(x=3,height=100*im/nm,bottom=0,color=cb['orangenature'],width=.7)
        ax.bar(x=3,height=100*em/nm,bottom=100*im/nm,color=cb['green'],width=.7)
    
    if nmh>0:
        ax.bar(x=4,height=100*imh/nmh,bottom=0,color=cb['orangenature'],width=.7)
        ax.bar(x=4,height=100*emh/nmh,bottom=100*imh/nmh,color=cb['green'],width=.7)
    
    if nh>0:
        ax.bar(x=5,height=100*ih/nh,bottom=0,color=cb['orangenature'],width=.7)
        ax.bar(x=5,height=100*eh/nh,bottom=100*ih/nh,color=cb['green'],width=.7)
    
    
    ax.set_ylabel('% Enhancers')
    
    ax.set_xlim(-.5,5.5)
    
    ax.set_xticks([0,1,2,3,4,5],['all',low,lm,med,mh,high])
    
    ax.set_xlabel('')
    ax.set_title('simulated data')

    plt.show()
    
    iter2bin2percentEnsDiffAndPvalue[iteration]={}
    
    for bi in REALDATA_bin2percentfunc:
        RA=REALDATA_bin2percentfunc[bi][0]
        Rtot=REALDATA_bin2percentfunc[bi][1]
        RI=Rtot-RA
    
        RApercent=100*RA/Rtot
        
        PA=PERMDATA_bin2percentfunc[bi][0]
        Ptot=PERMDATA_bin2percentfunc[bi][1]
        PI=Ptot-PA
    
        if Ptot==0: continue
        PApercent=100*PA/Ptot
    
        oddsratio, pvalue = stats.fisher_exact([[RA, PA], [RI, PI]])
    
        iter2bin2percentEnsDiffAndPvalue[iteration][bi]=(RApercent-PApercent,pvalue)
        
```

```python
fig,ax=quickfig()

ax.scatter(0,0,color='black')
xlabs=[]
for xi,bi in enumerate([low,lm,med,mh,high]):
    xlabs.append(bi)
    xi+=1

    if bi not in iter2bin2percentEnsDiffAndPvalue[iteration]: continue
    effects=[iter2bin2percentEnsDiffAndPvalue[iteration][bi][0] for iter in range(numits)]
    pvals=[iter2bin2percentEnsDiffAndPvalue[iteration][bi][1] for iter in range(numits)]

    maxpval=max(pvals)

    effectMean=np.mean(effects)
    effectSem  =stats.sem(effects)
    effectSd   =np.std(effects)

    ax.scatter(xi,effectMean,color='black')

ax.set_xticks(range(len(xlabs)+1),['all']+xlabs)
ax.axhline(0,ls='--',color='grey',zorder=-100)
ax.set_ylabel('∆ increase in active enhancers over null')
```
